```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python

DATA_PATH = Path("C:\\Users\\anush\\Customer-Journey-Reconstruction-Engine\\online_retail_II.xlsx")

year_1 = pd.read_excel(
    DATA_PATH,
    sheet_name="Year 2009-2010"
)

year_2 = pd.read_excel(
    DATA_PATH,
    sheet_name="Year 2010-2011"
)

year_1["SourceYear"] = "2009-2010"
year_2["SourceYear"] = "2010-2011"

df = pd.concat(
    [year_1, year_2],
    ignore_index=True
)

print("Combined shape:", df.shape)
print(df.head())
```

```python
print("Data types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isna().sum())

print("\nExact duplicate rows:", df.duplicated().sum())

print("\nDate range:")
print(df["InvoiceDate"].min())
print(df["InvoiceDate"].max())

print("\nUnique customers:",
      df["Customer ID"].nunique())

print("Unique invoices:", df["Invoice"].nunique())
print("Unique products:", df["StockCode"].nunique())
print("Countries:", df["Country"].nunique())
```

```python
invoice = df["Invoice"].astype(str)

print("Cancellation invoice rows:",
      invoice.str.startswith("C").sum())

print("Negative quantities:",
      (df["Quantity"] < 0).sum())

print("Zero prices:",
      (df["Price"] == 0).sum())

print("Negative prices:",
      (df["Price"] < 0).sum())

print("\nExamples of negative quantities:")
print(
    df.loc[df["Quantity"] < 0]
      .head(10)
)
```
